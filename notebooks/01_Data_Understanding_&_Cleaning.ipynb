The dataset consists of 7,043 customer records and 21 attributes. No duplicate records or explicit null values were detected. However, the TotalCharges column is stored as an object due to hidden blank entries that require preprocessing. The target variable, Churn, exhibits moderate class imbalance (73.46% non-churn vs. 26.54% churn), indicating that evaluation metrics beyond accuracy should be used during model development.

That blank value is not NaN; it's an empty string (or spaces). Since DECIMAL expects a numeric value, MySQL would reject it.

```python
df["TotalCharges"] = df["TotalCharges"].str.strip()
```

```python
import numpy as np
df["TotalCharges"] = df["TotalCharges"].replace("", np.nan)
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
```

```python
df["TotalCharges"].info()
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df[df["TotalCharges"].isnull()]
```

Because a customer who has just joined the company has paid nothing yet. So filling those rows with 0.

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df["TotalCharges"].isnull().sum()
```
